## Fonctions

Définition de functions

````
def is_odd(a):
    return a % 2 == 0

is_odd(2)
````

On peut définir des fonctions dans des fonctions

````
def is_odd(a):

    def is_divisible(number, base):
        return number % base == 0

    return is_divisible(a, 2)

is_odd(2)
is_divisible(6, 5)
````

Les arguments des fonctions peuvent avoir des valeurs par défaut

````
def add_or_subtract(first, second, operation="somme"):
    if operation == "somme":
        return first + second
    elif operation == "soustraction":
        return first - second
    else:
        print("Opération non autorisée")

add_or_subtract(first=3, second=4, operation="somme")
````

Varargs: arguments de taille variable

````
def sum_all(*args):
    # args is a list of arguments
    result = 0
    for arg in args:
        result += arg
    return result

# Call vararg function
print("Somme tous les entiers jusqu'à 10 =", sum_all(1, 2, 3, 4, 5, 6, 7, 8, 9))
````

Arguments nommées (Keyword arguments)

````
def print_pairs(**kwargs):
    # kwargs is a map
    for k, v in kwargs.items():
        print(k, v)

print_pairs(a=1, b=2)
````

Arguments nommés seulement

````
def create_car(*, speed, size):
    print("Voiture crée avec vitesse", speed, "et taille", size)

create_car(speed=9, size=3)
````

Fonctions comme paramètres

Il est possible de passer une fonction comme argument, *operation* ici est une fonction.

````
def reduce(array, operation):
    result = 0
    for k, v in enumerate(array):
        if k == 0:
            result = v
        else:
            result = operation(v, result)

    return result
````

Appliquer la fonction avec une autre fonction `add_or_subtract`


````
one_to_nine = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print(one_to_nine)

print("Somme sur le vecteur")
# L'opération est déduite du paramètre par défaut de add_or_substrant
reduce(one_to_nine, add_or_subtract)
````

Lambdas

Une fonction peut également être définie sans nom

````
print("Produit des éléments d'un vecteur")
reduce(one_to_nine, lambda x, y : x * y)
````

Fermetures (closures)

Une fonction peut retourner une autre fonction avec des comportements spécifiques selon les arguments.

````
def get_loss(op_reduce, op_foreach):

    def loss(a, b):
        c = []
        for av, bv in zip(a, b):
            c.append(op_foreach(av, bv))
        return op_reduce(c)

    return loss
````

Cette fonction peut servir à calculer l'erreur quadratique moyenne

````
mse_loss = get_loss(lambda x : sum(x) / len(x),
                    lambda a, b : (a - b) ** 2)
````

Ou l'erreur absolue moyenne (MAE)

````
mae_loss = get_loss(lambda x : sum(x) / len(x),
                    lambda a, b : abs(a - b))
````

On peut vérifier:

````
list1 = [0, 1, 1, 3, 0, 2, 3]
list2 = [1, 1, 2, 0, 0, 2, 3]

list_mse = mse_loss(list1, list2)
list_mae = mae_loss(list1, list2)

print("Deux listes:\n", list1, "\n", list2)
print("MSE Loss: {}\nMAE Loss: {}".format(list_mse, list_mae))
````
