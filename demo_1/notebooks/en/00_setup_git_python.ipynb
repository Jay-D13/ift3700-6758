## C. Git in ten minutes
Git records versions on your machine. GitHub hosts remote repositories.

**Working files → staging area → commit history** correspond to editing a file, `git add`, and `git commit`. A commit stays local until you push it to a remote.

Run the next cell in Colab. It creates a disposable practice repository and uses a clearly fictional identity only in that repository. Each run creates a fresh folder. No account or remote is required. Locally, this shell cell requires Git and Bash. On Windows, use Colab for this Git cell unless Bash is available to Jupyter.

Read the `status` output before and after staging. The second commit records one changed line.

```python
%%bash
set -eu
practice_dir=$(mktemp -d)
cd "$practice_dir"
git init -q
git config user.name "Demo Student"
git config user.email "demo@example.invalid"
printf 'My data science notes\n' > notes.txt
git status --short
git add notes.txt
git status --short
git commit -q -m "Add practice notes"
printf 'I can rerun a notebook.\n' >> notes.txt
git diff
git add notes.txt
git commit -q -m "Record notebook workflow"
git log --oneline -2
```
