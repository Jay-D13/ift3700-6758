## C. Git en dix minutes
Git enregistre les versions sur votre machine. GitHub héberge des dépôts distants.

**Fichiers de travail → staging area → historique des commits** correspondent à la modification d’un fichier, à `git add`, puis à `git commit`. Un commit reste local tant que vous ne le poussez pas vers un dépôt distant.

Exécutez la cellule suivante dans Colab. Elle crée un dépôt d’entraînement jetable et utilise une identité clairement fictive, uniquement dans ce dépôt. Chaque exécution crée un nouveau dossier. Aucun compte ni dépôt distant n’est nécessaire. En local, cette cellule shell nécessite Git et Bash. Sous Windows, utilisez Colab pour cette cellule Git, sauf si Bash est accessible à Jupyter.

Lisez la sortie de `status` avant et après le staging. Le deuxième commit enregistre la modification d’une ligne.

```python
%%bash
set -eu
practice_dir=$(mktemp -d)
cd "$practice_dir"
git init -q
git config user.name "Demo Student"
git config user.email "demo@example.invalid"
printf 'My data science notes\n' > notes.txt
git status --short
git add notes.txt
git status --short
git commit -q -m "Add practice notes"
printf 'I can rerun a notebook.\n' >> notes.txt
git diff
git add notes.txt
git commit -q -m "Record notebook workflow"
git log --oneline -2
```
